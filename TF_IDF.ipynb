```python
import pandas as pd

file_path = '/content/drive/MyDrive/Tugas/ETS ML 2/data_preprocessed.csv'
df = pd.read_csv(file_path)

# Display the first few rows of the dataframe
display(df.head())
```

```python

from sklearn.feature_extraction.text import TfidfVectorizer

# Mengambil teks yang sudah dipreprocessing
teks = df['clean_text'].fillna('')

# Membuat TF-IDF
tfidf = TfidfVectorizer(
    max_features=5000,
    ngram_range=(1, 2),
    sublinear_tf=True
)

X_tfidf = tfidf.fit_transform(teks)

print("Jumlah data:", X_tfidf.shape[0])
print("Jumlah fitur:", X_tfidf.shape[1])
print("Dimensi matriks TF-IDF:", X_tfidf.shape)
```

```python
import os
import pandas as pd

# Folder penyimpanan di Google Drive
folder = '/content/drive/MyDrive/Tugas/ETS ML 2/Hasil_TFIDF'
os.makedirs(folder, exist_ok=True)

file_csv = os.path.join(folder, 'data_tfidf.csv')
fitur = tfidf.get_feature_names_out()

# Simpan seluruh data secara bertahap
ukuran_batch = 100

for awal in range(0, X_tfidf.shape[0], ukuran_batch):
    akhir = min(awal + ukuran_batch, X_tfidf.shape[0])

    # Mengubah matriks menjadi tabel untuk setiap batch
    batch = pd.DataFrame(
        X_tfidf[awal:akhir].toarray(),
        columns=fitur
    )

    # Menambahkan label pada setiap baris
    batch.insert(
        0,
        'label_annotator',
        df['label_annotator'].iloc[awal:akhir].values
    )

    # Menyimpan semua batch ke satu file CSV
    batch.to_csv(
        file_csv,
        mode='w' if awal == 0 else 'a',
        header=(awal == 0),
        index=False
    )

    print(f"Progres: {akhir} dari {X_tfidf.shape[0]} data")

print("\nCSV lengkap berhasil disimpan!")
print("Lokasi:", file_csv)
print("Total baris data:", X_tfidf.shape[0])
print("Total fitur TF-IDF:", X_tfidf.shape[1])
print("Total kolom CSV:", X_tfidf.shape[1] + 1)
```
